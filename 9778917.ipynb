```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re

# IDS
# 1. load pull_request table
pr_df = pd.read_parquet('/home/aistudio/work/pull_request.parquet')

# 2. load comment tables
pr_comments = pd.read_parquet('/home/aistudio/work/pr_comments.parquet')

pr_review_comments = pd.read_parquet('/home/aistudio/work/pr_review_comments.parquet')


# 1. 提取 review_comments 中的 PR id
def extract_pr_id(url):
    if url is None:
        return None
    match = re.search(r'/pulls/(\d+)', url)
    if match:
        return int(match.group(1))
    return None

pr_review_comments["pr_id"] = pr_review_comments["pull_request_url"].apply(extract_pr_id)

# 2. 评论计数
pr_comment_count = pr_comments.groupby("pr_id").size().rename("pr_comment_count")

review_comment_count = pr_review_comments.groupby("pr_id").size().rename("review_comment_count")

# 3. merge 到 PR 表
pr_df = pr_df.merge(pr_comment_count, left_on="id", right_index=True, how="left")
pr_df = pr_df.merge(review_comment_count, left_on="id", right_index=True, how="left")
pr_df = pr_df.fillna({"pr_comment_count": 0, "review_comment_count": 0})

# 4. 计算每个 author 的 IDS
result = (
    pr_df.groupby("user_id")
    .apply(lambda g: (g["pr_comment_count"] + g["review_comment_count"]).sum() / len(g))
    .reset_index(name="IDS")
)

result


# tbs
# pr_df = pd.read_parquet('/home/aistudio/work/pull_request.parquet')

def compute_tbs(pr_df, user_col="user_id", time_col="created_at", min_prs_for_report=1):
    """
    Compute TBS per author from a pull-request DataFrame.
    pr_df: pandas.DataFrame with at least columns [user_col, time_col]
    user_col: column name for PR author id
    time_col: column name for PR created timestamp
    min_prs_for_report: optional filter to keep only users with >= this PR count
    Returns: DataFrame with columns [user_id, pr_count, first_pr, last_pr, active_days, TBS]
    """
    # 1. ensure timestamp dtype
    pr_df = pr_df.copy()
    pr_df[time_col] = pd.to_datetime(pr_df[time_col], errors="coerce")

    # drop PRs with missing author or time
    pr_df = pr_df.dropna(subset=[user_col, time_col])

    # 2. aggregate per author
    agg = pr_df.groupby(user_col).agg(
        pr_count = (time_col, "size"),
        first_pr = (time_col, "min"),
        last_pr  = (time_col, "max")
    ).reset_index()

    # 3. compute active days (days between first and last PR) + 1
    # use .dt.days to get integer days; ensure at least 1
    agg["active_days"] = (agg["last_pr"] - agg["first_pr"]).dt.days + 1
    agg["active_days"] = agg["active_days"].clip(lower=1)

    # 4. compute TBS
    agg["TBS"] = agg["pr_count"] / agg["active_days"]

    # 5. optional filter: keep only authors with >= min_prs_for_report
    if min_prs_for_report > 1:
        agg = agg[agg["pr_count"] >= min_prs_for_report].reset_index(drop=True)

    return agg

ai_tbs_df = compute_tbs(pr_df, user_col="user_id", time_col="created_at")

merged = result.merge(ai_tbs_df, on="user_id", how="left")
merged[['user_id','pr_count','IDS','TBS']].sort_values(
    by=['IDS','TBS'], ascending=[True, False]
).head(20)


# rds
def compute_rds(pr_df, pr_comments, pr_review_comments):
    

    # --- Step 1: parse PR id from pr_review_comments ---
    def extract_pr_id(url):
        if pd.isna(url):
            return None
        m = re.search(r'/pulls/(\d+)', url)
        return int(m.group(1)) if m else None

    review = pr_review_comments.copy()
    review["pr_id"] = review["pull_request_url"].apply(extract_pr_id)

    # --- Step 2: unify reviewer identity ---
    # pr_comments：reviewer 在 user_id 字段
    pc = pr_comments[["pr_id", "user_id"]].dropna()
    pc = pc.rename(columns={"user_id": "reviewer_id"})

    # pr_review_comments：reviewer 在 user 字段
    rc = review[["pr_id", "user"]].dropna()
    rc = rc.rename(columns={"user": "reviewer_id"})

    # 合并两类 reviewer
    all_reviews = pd.concat([pc, rc], ignore_index=True)

    # --- Step 3: PR → reviewers list ---
    pr_to_reviewers = all_reviews.groupby("pr_id")["reviewer_id"].apply(list)

    # --- Step 4: compute RDS per PR author ---
    result = []

    for user_id, group in pr_df.groupby("user_id"):
        pr_ids = group["id"].tolist()

        reviewers = []
        for pid in pr_ids:
            if pid in pr_to_reviewers:
                reviewers.extend(pr_to_reviewers[pid])

        total_reviews = len(reviewers)
        unique_reviewers = len(set(reviewers))

        rds = (unique_reviewers / total_reviews) if total_reviews > 0 else 0.0

        result.append({
            "user_id": user_id,
            "RDS": rds,
            "total_reviews": total_reviews,
            "unique_reviewers": unique_reviewers
        })

    return pd.DataFrame(result)

merged = result.merge(ai_tbs_df, on="user_id", how="left")
merged = merged.merge(rds_df, on="user_id", how="left")

len(merged)
#merged.head()
#len(merged)



# -----------------------------------------------------------
# 最终 Lone Wolf Score (LWS) 计算模块
# merged 表必须包含: IDS, TBS, RDS
# -----------------------------------------------------------

def compute_lws(merged, w1=0.4, w2=0.4, w3=0.2):
    df = merged.copy()

    # ---- Step 1: Z-score 标准化（避免不同尺度影响综合得分） ----
    # 避免标准差为 0 导致除零错误
    def zscore(series):
        std = series.std()
        if std == 0:
            return (series - series.mean())  # 全 0
        return (series - series.mean()) / std

    df["IDS_z"] = zscore(df["IDS"])
    df["TBS_z"] = zscore(df["TBS"])
    df["RDS_z"] = zscore(df["RDS"])

    # ---- Step 2: Lone Wolf 转换方向 ----
    # IDS 越低越孤立 → (1 - IDS_z_norm)
    # TBS 越高越孤立 → 直接用 TBS_z
    # RDS 越低越孤立 → (1 - RDS_z_norm)

    # 为了让 (1 - z) 有意义，把 Z-score 拉到 [0,1] 范围
    def min_max_norm(col):
        mn = col.min()
        mx = col.max()
        if mx - mn == 0:
            return col * 0
        return (col - mn) / (mx - mn)

    df["IDS_norm"] = min_max_norm(df["IDS_z"])
    df["TBS_norm"] = min_max_norm(df["TBS_z"])
    df["RDS_norm"] = min_max_norm(df["RDS_z"])

    # ---- Step 3: 构建 Lone Wolf Score ----
    df["LWS"] = (
        w1 * (1 - df["IDS_norm"]) +
        w2 * df["TBS_norm"] +
        w3 * (1 - df["RDS_norm"])
    )

    return df

#ai_lws_df.to_parquet("ai_lws_df.parquet")

# -----------------------------------------------------------
# 使用示例（假设你已经合并 IDS + TBS + RDS 得到 merged）
# -----------------------------------------------------------

lws_df = compute_lws(merged)

# 查看得分最高（最孤狼）的前 5 名

print("Check out the top 5 with the highest scores (the most lone wolves)")
top_lone_wolves = lws_df.sort_values("LWS", ascending=False).head()
print(top_lone_wolves[["user_id", "IDS", "TBS", "RDS", "LWS"]])

# 基于分位数
#q75 = lws_df["LWS"].quantile(0.75)

#lws_df["LoneWolf_flag"] = (lws_df["LWS"] >= q75).astype(int)

#print("LWS 阈值 =", q75)

#print(lws_df[["user_id", "LWS", "LoneWolf_flag"]].head(10))





# Example: pick the top lone wolf
user_lone = lws_df.sort_values("LWS", ascending=False).iloc[0]
user_collab = lws_df.sort_values("LWS", ascending=True).iloc[0]

print(user_lone)
print("-----------------------------------------------")
print(user_collab)
# 雷达图的三个指标
labels = ["IDS", "TBS", "RDS"]

# Lone Wolf 的数值
values_lone = [user_lone["IDS"], user_lone["TBS"], user_lone["RDS"]]
values_lone += values_lone[:1]  # 闭合

# Collaborative 的数值
values_collab = [user_collab["IDS"], user_collab["TBS"], user_collab["RDS"]]
values_collab += values_collab[:1]  # 闭合

# 角度设置
angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]

# -------------------------------------------------------
# 2. 画图（两个雷达图并排）
# -------------------------------------------------------
fig, ax = plt.subplots(1, 2, subplot_kw=dict(polar=True), figsize=(14, 6))

# --- 图 1：最 Lone Wolf ---
ax[0].plot(angles, values_lone, marker='o')
ax[0].fill(angles, values_lone, alpha=0.25)
ax[0].set_xticks(angles[:-1])
ax[0].set_xticklabels(labels)
ax[0].set_title(f"Lone Wolf User (user_id={user_lone['user_id']})")

# --- 图 2：最合作用户 ---
ax[1].plot(angles, values_collab, marker='o')
ax[1].fill(angles, values_collab, alpha=0.25)
ax[1].set_xticks(angles[:-1])
ax[1].set_xticklabels(labels)
ax[1].set_title(f"Collaborative User (user_id={user_collab['user_id']})")

plt.tight_layout()
plt.show()

ai_lws_df =lws_df
ai_lws_df.to_parquet("ai_lws_df.parquet")
```

```python
import pandas as pd
import re
import matplotlib.pyplot as plt
import numpy as np


# human PR
human_pr = pd.read_parquet('/home/aistudio/work/human_pull_request.parquet')

# 本来 AI 用的是 pr_comments / pr_review_comments，Human 时间相同
pr_comments = pd.read_parquet('/home/aistudio/work/pr_comments.parquet')
pr_review_comments = pd.read_parquet('/home/aistudio/work/pr_review_comments.parquet')



#Human IDS 完整 Python 代码
def extract_pr_number(url):
    if pd.isna(url):
        return None
    m = re.search(r'/pulls/(\d+)', url)
    return int(m.group(1)) if m else None

pr_review_comments = pr_review_comments.copy()
pr_review_comments["pr_number"] = pr_review_comments["pull_request_url"].apply(extract_pr_number)

review_comment_count = (
    pr_review_comments
    .dropna(subset=["pr_number"])
    .groupby("pr_number")
    .size()
    .rename("review_comment_count")
)

human_pr = human_pr.copy()

human_pr = human_pr.merge(
    review_comment_count.to_frame(),
    how="left",
    left_on="number",
    right_index=True
)

human_pr["review_comment_count"] = (
    human_pr["review_comment_count"].fillna(0).astype(int)
)

ids_list = []
for uid, g in human_pr.groupby("user_id"):
    Vol = len(g)
    total_interactions = g["review_comment_count"].sum()
    IDS = total_interactions / Vol if Vol > 0 else 0
    ids_list.append({"user_id": uid, "pr_count": Vol, "IDS": IDS})

ids_df_human = pd.DataFrame(ids_list)

#print(ids_df_human.head())

#ids_df_human.head()
#ids_df_human.sort_values("IDS", ascending=False).head(10)


#Human TBS 完整 Python 代码

def compute_tbs(human_pr, user_col="user_id", time_col="created_at", min_prs_for_report=1):
    """
    Compute TBS for human PR submitters.
    """
    df = human_pr.copy()

    # 1. timestamp 转换
    df[time_col] = pd.to_datetime(df[time_col], errors="coerce")

    # 去掉缺失数据
    df = df.dropna(subset=[user_col, time_col])

    # 2. 聚合
    agg = df.groupby(user_col).agg(
        pr_count=(time_col, "size"),
        first_pr=(time_col, "min"),
        last_pr=(time_col, "max"),
    ).reset_index()

    # 3. active_days 计算
    agg["active_days"] = (agg["last_pr"] - agg["first_pr"]).dt.days + 1
    agg["active_days"] = agg["active_days"].clip(lower=1)

    # 4. TBS
    agg["TBS"] = agg["pr_count"] / agg["active_days"]

    # 可选：要求至少 min_prs_for_report
    if min_prs_for_report > 1:
        agg = agg[agg["pr_count"] >= min_prs_for_report].reset_index(drop=True)

    return agg

# ---------------------------
# 使用示例
# ---------------------------
human_tbs_df = compute_tbs(human_pr)

#print(human_tbs_df.head(20))  # 查看前几行






# 合并 IDS 与 TBS，按 user_id
human_merged = ids_df_human.merge(
    human_tbs_df[["user_id", "TBS"]],
    on="user_id",
    how="left"
)

print(human_merged.head())
print(human_merged.shape)

# 按 IDS 从低到高（独狼更明显），再按 TBS 从高到低排序
human_merged.sort_values(
    by=["IDS", "TBS"],
    ascending=[True, False]
).head(10)


print(human_merged.columns)

# Step 2: 仅保留 reviewer（使用 "user" 而不是 user_id）
rc = pr_review_comments[["pr_number", "user"]].dropna()
rc = rc.rename(columns={"user": "reviewer"})   # reviewer 名称（字符串）



# Step 3: 为每个 PR 构建 reviewer 列表
pr_to_reviewers = rc.groupby("pr_number")["reviewer"].apply(list)

# Step 4: 计算 RDS per human author
rds_list = []

for uid, g in human_pr.groupby("user_id"):
    pr_numbers = g["number"].tolist()

    reviewers = []
    for prnum in pr_numbers:
        if prnum in pr_to_reviewers:
            reviewers.extend(pr_to_reviewers[prnum])

    total_reviews = len(reviewers)
    unique_reviewers = len(set(reviewers))

    if total_reviews == 0:
        RDS = 0.0
    else:
        RDS = unique_reviewers / total_reviews

    rds_list.append({
        "user_id": uid,
        "total_reviews": total_reviews,
        "unique_reviewers": unique_reviewers,
        "RDS": RDS
    })

rds_df_human = pd.DataFrame(rds_list)

#print(rds_df_human.head())
#print(rds_df_human.shape)




# Step 2: Merge IDS + TBS + RDS for human contributors

# 1. 先合并 IDS + TBS
human_features = ids_df_human.merge(
    human_tbs_df[["user_id", "TBS"]],
    on="user_id",
    how="left"
)

# 2. 再合并 RDS
human_features = human_features.merge(
    rds_df_human[["user_id", "RDS", "total_reviews", "unique_reviewers"]],
    on="user_id",
    how="left"
)

# 3. 缺失值填 0（部分用户可能没有 review 评论）
human_features[["TBS", "RDS", "total_reviews", "unique_reviewers"]] = (
    human_features[["TBS", "RDS", "total_reviews", "unique_reviewers"]].fillna(0)
)


print("-----------------------")
print(human_features.head())
print(human_features.shape)

print("******************************")

# 复制避免修改原表
lws_df = human_features.copy()

# Step 1: Min-Max normalize three metrics: IDS, TBS, RDS
def min_max_norm(series):
    if series.max() == series.min():
        return series * 0  # 全相等则直接全部归 0
    return (series - series.min()) / (series.max() - series.min())

lws_df["IDS_norm"] = min_max_norm(lws_df["IDS"])
lws_df["TBS_norm"] = min_max_norm(lws_df["TBS"])
lws_df["RDS_norm"] = min_max_norm(lws_df["RDS"])

# Step 2: 计算最终 Lone Wolf Score (LWS)
w1, w2, w3 = 0.4, 0.4, 0.2

lws_df["LWS"] = (
    w1 * (1 - lws_df["IDS_norm"]) +   # IDS 越低 → 越 Lone Wolf
    w2 * (lws_df["TBS_norm"]) +       # TBS 越高 → 越 Lone Wolf
    w3 * (1 - lws_df["RDS_norm"])     # RDS 越低 → 越 Lone Wolf
)

# Step 3: 排序查看最 Lone Wolf 的前 10 个用户
top_lone_wolves = lws_df.sort_values("LWS", ascending=False).head(10)

print(top_lone_wolves)
print("Total users:", len(lws_df))
lws_df.sort_values("LWS", ascending=False).head(10)



# -------------------------
# 1. 选择 top1 与 bottom1
# -------------------------
top1 = lws_df.sort_values("LWS", ascending=False).iloc[0]
bottom1 = lws_df.sort_values("LWS", ascending=True).iloc[0]

# 雷达图指标（都是 0-1 之间）
labels = ["IDS_norm", "TBS_norm", "RDS_norm"]

top_vals = [top1["IDS_norm"], top1["TBS_norm"], top1["RDS_norm"]]
bottom_vals = [bottom1["IDS_norm"], bottom1["TBS_norm"], bottom1["RDS_norm"]]

# 闭合雷达图（结尾再加上第一个）
top_vals += top_vals[:1]
bottom_vals += bottom_vals[:1]

# 角度
angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]

# -------------------------
# 2. 绘图
# -------------------------
plt.figure(figsize=(7, 7))
ax = plt.subplot(111, polar=True)

# 绘制 top1
ax.plot(angles, top_vals, linewidth=2, label="Top-1 Lone Wolf")
ax.fill(angles, top_vals, alpha=0.25)

# 绘制 bottom1
ax.plot(angles, bottom_vals, linewidth=2, label="Bottom-1 Lone Wolf")
ax.fill(angles, bottom_vals, alpha=0.25)

# 设置标签
ax.set_xticks(angles[:-1])
ax.set_xticklabels(["IDS (1-IDS)", "TBS", "RDS (1-RDS)"])

# 标题
plt.title("Human Lone Wolf Radar Chart\nTop-1 vs Bottom-1", fontsize=14)

# 图例（放在底部稍微偏右）
plt.legend(loc="upper right", bbox_to_anchor=(1.2, 1.1))

plt.tight_layout()
plt.show()

human_lws_df=lws_df
human_lws_df.to_parquet("human_lws_df.parquet")
```

```python
import pandas as pd
import numpy as np

# ==============================================================================
# 1. 数据加载与标准化 (Load & Normalize Data)
# ==============================================================================
print(">>> [Step 1] 正在加载 Parquet 数据...")

# --- A. 加载 PR 表 ---
# 假设 AI PR 表列名包含: id, user, repo_url, created_at
df_ai_pr = pd.read_parquet('/home/aistudio/work/pull_request.parquet')

# 假设 Human PR 表列名包含: id, user, repo_url, created_at
df_human_pr = pd.read_parquet('/home/aistudio/work/human_pull_request.parquet')

# 标准化列名：统一将主键重命名为 'pr_id'，确保包含 'repo_url'
def standardize_pr_columns(df, tag):
    # 自动识别 ID 列
    id_col = 'id' if 'id' in df.columns else 'pr_id'
    
    # 必要的列 (请根据实际 Parquet 文件微调列名)
    req_cols = [id_col, 'user', 'repo_url', 'created_at']
    
    # 检查列是否存在
    missing = [c for c in req_cols if c not in df.columns]
    if missing:
        raise ValueError(f"{tag} 表缺失必要列: {missing}")
        
    # 筛选并重命名
    df_clean = df[req_cols].rename(columns={id_col: 'pr_id'})
    # 强制转换时间为 UTC，避免时区错误
    df_clean['created_at'] = pd.to_datetime(df_clean['created_at'], utc=True)
    return df_clean

try:
    df_ai_clean = standardize_pr_columns(df_ai_pr, "AI PR")
    df_human_clean = standardize_pr_columns(df_human_pr, "Human PR")
except Exception as e:
    print(f"数据预处理错误: {e}")
    print("请检查 Parquet 文件的列名，确保包含 id, user, repo_url, created_at")
    exit()

print(f"    -> AI PR 加载: {len(df_ai_clean)} 条")
print(f"    -> Human PR 加载: {len(df_human_clean)} 条")

# --- B. 加载评论表 (双表合并) ---
print(">>> [Step 2] 加载并合并评论数据...")

# 读取两个评论表
try:
    df_comments = pd.read_parquet('/home/aistudio/work/pr_comments.parquet')
    df_reviews = pd.read_parquet('/home/aistudio/work/pr_review_comments.parquet')
except FileNotFoundError as e:
    print(f"错误: 找不到评论文件 ({e})，请确保文件在当前目录下。")
    exit()

# 提取 PR ID 的辅助函数
def extract_pr_ids(df, source_name):
    # 常见的 ID 列名
    possible_cols = ['pull_request_id', 'pr_id', 'issue_id', 'id']
    for col in possible_cols:
        if col in df.columns:
            return set(df[col].dropna().unique())
    print(f"警告: 在 {source_name} 中未找到关联 PR 的 ID 列。")
    return set()

ids_general = extract_pr_ids(df_comments, "pr_comments")
ids_review = extract_pr_ids(df_reviews, "pr_review_comments")

# 取并集：只要在任意表出现，就算"有沟通"
communicated_pr_ids = ids_general.union(ids_review)

print(f"    -> 通用评论涉及 PR 数: {len(ids_general)}")
print(f"    -> 审查评论涉及 PR 数: {len(ids_review)}")
print(f"    -> 合并后有沟通的 PR 总数: {len(communicated_pr_ids)}")

# ==============================================================================
# 2. 协作环境判定 (Collaboration Detection - Repo Level)
# ==============================================================================
print("\n>>> [Step 3] 正在判定协作环境 (寻找活跃的人类仓库)...")

# 逻辑：将 AI PR 和 Human PR 按 'repo_url' 连接
# 这一步会产生大量数据，如果是超大数据集，建议分块处理
merged = pd.merge(
    df_ai_clean,
    df_human_clean,
    on='repo_url',
    suffixes=('_ai', '_human')
)

# 计算时间差 (绝对值天数)
merged['days_diff'] = (merged['created_at_ai'] - merged['created_at_human']).dt.days.abs()

# 筛选：±90天内有人类活动的仓库
TIME_WINDOW = 90
active_collab_env = merged[merged['days_diff'] <= TIME_WINDOW]

# 提取处于"活跃环境"中的 AI PR ID (去重)
collaborating_ai_pr_ids = active_collab_env['pr_id_ai'].unique()

print(f"    -> 发现 {len(collaborating_ai_pr_ids)} 个 AI PR 处于活跃的人类协作仓库中。")

# ==============================================================================
# 3. 独狼判定 (Lone Wolf Detection)
# ==============================================================================
print("\n>>> [Step 4] 正在筛选 Lone Wolf (协作但无沟通)...")

results = []

for pr_id in collaborating_ai_pr_ids:
    # 核心判断逻辑:
    # 1. 它是协作 PR (在 active_collab_env 中) -> 满足
    # 2. 它不在沟通列表中 (communicated_pr_ids) -> 满足
    if pr_id not in communicated_pr_ids:
        
        # 获取 AI PR 信息
        ai_info = df_ai_clean[df_ai_clean['pr_id'] == pr_id].iloc[0]
        
        # 获取"共存证据" (找一个离得最近的人类 PR 作为证据)
        evidence = active_collab_env[active_collab_env['pr_id_ai'] == pr_id].sort_values('days_diff').iloc[0]
        
        results.append({
            'Lone_Wolf_PR_ID': pr_id,
            'Repo_URL': ai_info['repo_url'],
            'AI_Agent': ai_info['user'],
            'AI_Created_At': ai_info['created_at'].date(),
            'Nearest_Human_PR': evidence['pr_id_human'],
            'Human_User': evidence['user_human'],
            'Human_Created_At': evidence['created_at_human'].date(),
            'Gap_Days': evidence['days_diff']
        })

df_results = pd.DataFrame(results)

# ==============================================================================
# 4. 结果展示
# ==============================================================================
print("\n" + "="*80)
print(f"【检测报告】AI Lone Wolf (Repo-Level Heuristic)")
print(f"定义: AI 在活跃人类仓库(±{TIME_WINDOW}天)提交 PR，且无任何 General/Review 评论")
print("="*80)

if not df_results.empty:
    print(f"共发现 {len(df_results)} 个 Lone Wolf PR。")
    print("-" * 80)
    # 调整显示宽度
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    pd.set_option('display.max_colwidth', 40)
    
    # 显示前 20 条
    print(df_results.head(20).to_string(index=False))
    
    # 导出结果 (可选)
    # df_results.to_csv('ai_lone_wolves_result.csv', index=False)
    # print("\n完整结果已保存至 'ai_lone_wolves_result.csv'")
else:
    print("未检测到符合定义的 Lone Wolf 行为。")
```

```python
import pandas as pd
import numpy as np
import os

# ==============================================================================
# 1. 数据加载与标准化 (Load & Normalize Data)
# ==============================================================================
print(">>> [Step 1] 正在加载 Parquet 数据...")

# 检查文件是否存在
required_files = ['/home/aistudio/work/pull_request.parquet', '/home/aistudio/work/human_pull_request.parquet', 
                  '/home/aistudio/work/pr_comments.parquet', '/home/aistudio/work/pr_review_comments.parquet']
for f in required_files:
    if not os.path.exists(f):
        print(f"错误: 找不到文件 '{f}'。请确保所有必要文件都在当前目录下。")
        exit()

# --- A. 加载 PR 表 ---
df_ai_pr = pd.read_parquet('/home/aistudio/work/pull_request.parquet')
df_human_pr = pd.read_parquet('/home/aistudio/work/human_pull_request.parquet')

# 标准化列名：统一将主键重命名为 'pr_id'，并确保包含必要的列
def standardize_pr_columns(df, tag):
    # 自动识别 ID 列
    id_col = 'id' if 'id' in df.columns else 'pr_id'
    
    # 【关键修改】：这里加入了 'agent' 列，确保它不会被过滤掉
   
    req_cols = [id_col, 'user', 'repo_url', 'created_at', 'agent', 'state']
    
    # 检查列是否存在
    missing = [c for c in req_cols if c not in df.columns]
    if missing:
        # 如果Human表有agent列但AI表没有，或者反之，这里会报错提醒
        raise ValueError(f"{tag} 表缺失必要列: {missing}")
        
    # 筛选并重命名
    df_clean = df[req_cols].rename(columns={id_col: 'pr_id'})
    # 强制转换时间为 UTC，避免时区错误
    df_clean['created_at'] = pd.to_datetime(df_clean['created_at'], utc=True)
    return df_clean

try:
    print("    -> 正在标准化 AI PR 表...")
    df_ai_clean = standardize_pr_columns(df_ai_pr, "AI PR")
    print("    -> 正在标准化 Human PR 表...")
    df_human_clean = standardize_pr_columns(df_human_pr, "Human PR")
except Exception as e:
    print(f"\n数据预处理错误: {e}")
    print("请检查您的 Parquet 文件，确保它们都包含以下列：'id'(或'pr_id'), 'user', 'repo_url', 'created_at', 'agent'")
    exit()

print(f"    -> AI PR 加载完成: {len(df_ai_clean)} 条")
print(f"    -> Human PR 加载完成: {len(df_human_clean)} 条")

# --- B. 加载评论表 (双表合并) ---
print("\n>>> [Step 2] 加载并合并评论数据...")

df_comments = pd.read_parquet('/home/aistudio/work/pr_comments.parquet')
df_reviews = pd.read_parquet('/home/aistudio/work/pr_review_comments.parquet')

# 提取 PR ID 的辅助函数
def extract_pr_ids(df, source_name):
    # 常见的 ID 列名
    possible_cols = ['pull_request_id', 'pr_id', 'issue_id', 'id']
    for col in possible_cols:
        if col in df.columns:
            # 确保 ID 是统一的格式（例如字符串或整数），这里统一转为字符串以防万一
            return set(df[col].dropna().astype(str).unique())
    print(f"警告: 在 {source_name} 中未找到关联 PR 的 ID 列。")
    return set()

# 确保主表的 ID 格式也一致
df_ai_clean['pr_id'] = df_ai_clean['pr_id'].astype(str)
df_human_clean['pr_id'] = df_human_clean['pr_id'].astype(str)

ids_general = extract_pr_ids(df_comments, "pr_comments")
ids_review = extract_pr_ids(df_reviews, "pr_review_comments")

# 取并集：只要在任意表出现，就算"有沟通"
communicated_pr_ids = ids_general.union(ids_review)

print(f"    -> 通用评论涉及 PR 数: {len(ids_general)}")
print(f"    -> 审查评论涉及 PR 数: {len(ids_review)}")
print(f"    -> 合并后有沟通的 PR 总数: {len(communicated_pr_ids)}")

# ==============================================================================
# 2. 协作环境判定 (Collaboration Detection - Repo Level)
# ==============================================================================
print("\n>>> [Step 3] 正在判定协作环境 (寻找活跃的人类仓库)...")

# 逻辑：将 AI PR 和 Human PR 按 'repo_url' 连接
# 这会找出所有"AI和人类都在同一个仓库提交过代码"的记录
# 注意：如果数据量巨大，这一步可能会比较慢
merged = pd.merge(
    df_ai_clean,
    df_human_clean,
    on='repo_url',
    suffixes=('_ai', '_human')
)

# 计算时间差 (绝对值天数)
merged['days_diff'] = (merged['created_at_ai'] - merged['created_at_human']).dt.days.abs()

# 筛选：±90天内有人类活动的仓库
TIME_WINDOW = 90
active_collab_env = merged[merged['days_diff'] <= TIME_WINDOW]

# 提取处于"活跃环境"中的 AI PR ID (去重)
collaborating_ai_pr_ids = active_collab_env['pr_id_ai'].unique()

print(f"    -> 发现 {len(collaborating_ai_pr_ids)} 个 AI PR 处于活跃的人类协作仓库中(±{TIME_WINDOW}天)。")

# ==============================================================================
# 3. 独狼判定 (Lone Wolf Detection)
# ==============================================================================
print("\n>>> [Step 4] 正在筛选 Lone Wolf (协作但无沟通)...")

results = []

# 使用集合提高查找速度
collab_ai_pr_ids_set = set(collaborating_ai_pr_ids)

for pr_id in collab_ai_pr_ids_set:
    # 核心判断逻辑:
    # 1. 它是协作 PR (在 active_collab_env 中) -> 已满足
    # 2. 它不在沟通列表中 (communicated_pr_ids) -> 满足
    if pr_id not in communicated_pr_ids:
        
        # 获取 AI PR 信息
        ai_info = df_ai_clean[df_ai_clean['pr_id'] == pr_id].iloc[0]
        
        # 获取"共存证据" (找一个离得最近的人类 PR 作为证据)
        # 注意：这里 evidence['pr_id_human'] 可能有多个，取第一个即可
        evidence_df = active_collab_env[active_collab_env['pr_id_ai'] == pr_id]
        evidence = evidence_df.sort_values('days_diff').iloc[0]
        
        results.append({
            'Lone_Wolf_PR_ID': pr_id,
            'Repo_URL': ai_info['repo_url'],
            # 【关键修改】：这里现在可以使用 ai_row['agent'] 了
            'AI_Agent': ai_info['agent'], 
            'AI_User_Name': ai_info['user'], # 保留 user 以便对比
            'AI_Created_At': ai_info['created_at'].date(),
            # 对于人类，我们通常还是用 user 列来标识
            'Evidence_Human_User': evidence['user_human'],
            'Gap_Days': evidence['days_diff']
        })

df_results = pd.DataFrame(results)

# ==============================================================================
# 4. 结果展示
# ==============================================================================
print("\n" + "="*80)
print(f"【检测报告】AI Lone Wolf (仓库级协作判定)")
print(f"定义: AI 在活跃人类仓库提交 PR，且无任何 General/Review 评论")
print("="*80)

if not df_results.empty:
    print(f"共发现 {len(df_results)} 个 Lone Wolf PR。")
    print("-" * 80)
    # 调整显示宽度
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    pd.set_option('display.max_colwidth', 30)
    
    # 显示前 20 条，包含新的 AI_Agent 列
    print(df_results.head(20).to_string(index=False))
    
    # 导出结果 (可选)
    output_file = 'ai_lone_wolves_result.csv'
    # df_results.to_csv(output_file, index=False)
    # print(f"\n完整结果已保存至 '{output_file}'")
else:
    print("未检测到符合定义的 Lone Wolf 行为。")



#RQ2
print("以下是RQ2的代码。")

# ==============================================================================
# RQ2: 独狼行为的影响 (Impact Analysis)
# ==============================================================================
print("\n>>> 正在进行 RQ2 分析：独狼行为对 PR 合并率的影响...")

# 1. 确定研究对象：必须是处于"活跃人类仓库"中的 AI PR
# (我们不对比那些在死仓库里的 PR，因为没有可比性)
target_population = active_collab_env.copy()

# 2. 打标签：是否为 Lone Wolf
# 逻辑：如果 PR ID 不在沟通列表中，就是 Lone Wolf
target_population['is_lone_wolf'] = ~target_population['pr_id_ai'].isin(communicated_pr_ids)

# 3. 去重 (因为 merge 操作可能导致 PR 重复出现，我们只关心 PR 本身)
analysis_df = target_population[['pr_id_ai', 'state', 'is_lone_wolf']].drop_duplicates()

# 4. 计算合并率 (Merge Rate)
# 假设 state 列的值为 'merged', 'closed', 'open'
# 某些数据集用 merged_at 是否为空来判断，这里假设用 state
def calculate_metrics(df, group_name):
    total = len(df)
    if total == 0:
        return
    merged_count = len(df[df['state'] == 'merged'])
    merge_rate = (merged_count / total) * 100
    print(f"[{group_name}] 总数: {total} | 合并数: {merged_count} | 合并率: {merge_rate:.2f}%")

print("-" * 50)
print("对比结果：")
calculate_metrics(analysis_df[analysis_df['is_lone_wolf'] == True], "Group A: 独狼 AI")
calculate_metrics(analysis_df[analysis_df['is_lone_wolf'] == False], "Group B: 社交 AI")
print("-" * 50)

# 5. (可选进阶) 统计检验
# 如果你想让论文更严谨，可以用卡方检验 (Chi-square test) 看差异是否显著
from scipy.stats import chi2_contingency

group_a = analysis_df[analysis_df['is_lone_wolf'] == True]
group_b = analysis_df[analysis_df['is_lone_wolf'] == False]

if len(group_a) > 0 and len(group_b) > 0:
    merged_a = len(group_a[group_a['state'] == 'merged'])
    rejected_a = len(group_a) - merged_a
    merged_b = len(group_b[group_b['state'] == 'merged'])
    rejected_b = len(group_b) - merged_b

    # 构建列联表
    contingency = [[merged_a, rejected_a], [merged_b, rejected_b]]
    chi2, p, _, _ = chi2_contingency(contingency)
    print(f"统计检验 (Chi-square): p-value = {p:.4e}")
    if p < 0.05:
        print("结论：两组的合并率存在【统计学显著】差异。")
    else:
        print("结论：两组差异不显著。")
```

```python
import pandas as pd
import numpy as np
import os
from scipy.stats import chi2_contingency

# ==============================================================================
# [Step 1] 数据加载与标准化 (Load & Normalize Data)
# ==============================================================================
print(">>> [Step 1] 正在加载 Parquet 数据...")

required_files = ['/home/aistudio/work/pull_request.parquet', '/home/aistudio/work/human_pull_request.parquet', 
                  '/home/aistudio/work/pr_comments.parquet', '/home/aistudio/work/pr_review_comments.parquet']
for f in required_files:
    if not os.path.exists(f):
        print(f"错误: 找不到文件 '{f}'。请确保所有必要文件都在当前目录下。")
        exit()

df_ai_pr = pd.read_parquet('/home/aistudio/work/pull_request.parquet')
df_human_pr = pd.read_parquet('/home/aistudio/work/human_pull_request.parquet')

def standardize_pr_columns(df, tag):
    id_col = 'id' if 'id' in df.columns else 'pr_id'
    
    # 【关键修正 1】：加入 'merged_at' 列，确保它被加载
    # 我们需要 state 和 merged_at 共同判断合并状态
    req_cols = [id_col, 'user', 'repo_url', 'created_at', 'agent', 'state', 'merged_at']
    
    missing = [c for c in req_cols if c not in df.columns]
    if missing:
        print(f"警告: {tag} 表缺失列: {missing}。后续分析(尤其是RQ2)可能会受影响。")
        existing_cols = [c for c in req_cols if c in df.columns]
        df_clean = df[existing_cols].rename(columns={id_col: 'pr_id'})
    else:
        df_clean = df[req_cols].rename(columns={id_col: 'pr_id'})
        
    # 时间格式转换
    df_clean['created_at'] = pd.to_datetime(df_clean['created_at'], utc=True)
    # 如果 merged_at 存在，也转为时间格式以便判断 notna()
    if 'merged_at' in df_clean.columns:
        df_clean['merged_at'] = pd.to_datetime(df_clean['merged_at'], utc=True, errors='coerce')

    # 统一 state 的值格式 (转小写，去空格)
    if 'state' in df_clean.columns:
        df_clean['state'] = df_clean['state'].astype(str).str.lower().str.strip()
        
    return df_clean

try:
    print("    -> 正在标准化 AI PR 表...")
    df_ai_clean = standardize_pr_columns(df_ai_pr, "AI PR")
    print("    -> 正在标准化 Human PR 表...")
    # Human表可能没有agent列，这里允许缺失，只要保证AI表有即可
    df_human_clean = standardize_pr_columns(df_human_pr, "Human PR")
except Exception as e:
    print(f"\n数据预处理错误: {e}")
    exit()

print(f"    -> AI PR 加载完成: {len(df_ai_clean)} 条")
print(f"    -> Human PR 加载完成: {len(df_human_clean)} 条")

# --- 加载评论表 ---
print("\n>>> [Step 2] 加载并合并评论数据...")
df_comments = pd.read_parquet('/home/aistudio/work/pr_comments.parquet')
df_reviews = pd.read_parquet('/home/aistudio/work/pr_review_comments.parquet')

def extract_pr_ids(df, source_name):
    possible_cols = ['pull_request_id', 'pr_id', 'issue_id', 'id']
    for col in possible_cols:
        if col in df.columns:
            return set(df[col].dropna().astype(str).unique())
    print(f"警告: 在 {source_name} 中未找到关联 PR 的 ID 列。")
    return set()

df_ai_clean['pr_id'] = df_ai_clean['pr_id'].astype(str)
# df_human_clean['pr_id'] = df_human_clean['pr_id'].astype(str) # Human不需要

ids_general = extract_pr_ids(df_comments, "pr_comments")
ids_review = extract_pr_ids(df_reviews, "pr_review_comments")
communicated_pr_ids = ids_general.union(ids_review)
print(f"    -> 合并后有沟通的 PR 总数: {len(communicated_pr_ids)}")

# ==============================================================================
# [Step 3] 协作环境判定 (Repo Level)
# ==============================================================================
print("\n>>> [Step 3] 正在判定协作环境 (寻找活跃的人类仓库)...")
# 注意：Human表标准化时如果缺列可能导致后续merge报错，确保Human表至少有 repo_url 和 created_at
if 'repo_url' not in df_human_clean.columns or 'created_at' not in df_human_clean.columns:
      print("错误：Human PR 表缺少 repo_url 或 created_at，无法判定协作环境。")
      exit()

merged = pd.merge(
    df_ai_clean,
    df_human_clean,
    on='repo_url',
    suffixes=('_ai', '_human')
)
merged['days_diff'] = (merged['created_at_ai'] - merged['created_at_human']).dt.days.abs()
TIME_WINDOW = 90
active_collab_env = merged[merged['days_diff'] <= TIME_WINDOW]
collaborating_ai_pr_ids = active_collab_env['pr_id_ai'].unique()
print(f"    -> 发现 {len(collaborating_ai_pr_ids)} 个 AI PR 处于活跃的人类协作仓库中。")

# ==============================================================================
# RQ1: 独狼检测结果 (Lone Wolf Detection)
# ==============================================================================
print("\n" + "="*80)
print(">>> [RQ1 Analysis] Lone Wolf 检测报告")
print("="*80)

results = []
collab_ai_pr_ids_set = set(collaborating_ai_pr_ids)

for pr_id in collab_ai_pr_ids_set:
    if pr_id not in communicated_pr_ids:
        ai_info = df_ai_clean[df_ai_clean['pr_id'] == pr_id].iloc[0]
        # 取最近的一个证据
        evidence = active_collab_env[active_collab_env['pr_id_ai'] == pr_id].sort_values('days_diff').iloc[0]
        
        results.append({
            'Lone_Wolf_PR_ID': pr_id,
            'Repo_URL': ai_info['repo_url'],
            'AI_Agent': ai_info.get('agent', 'N/A'), 
            # 'State': ai_info.get('state', 'unknown'), # RQ1 展示可以省略 state
            'Evidence_Human': evidence['user_human'],
            'Gap_Days': evidence['days_diff']
        })

df_results_rq1 = pd.DataFrame(results)

if not df_results_rq1.empty:
    print(f"共发现 {len(df_results_rq1)} 个 Lone Wolf PR (占比: {len(df_results_rq1)/len(collaborating_ai_pr_ids)*100:.2f}% 的协作 PR)。")
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    print(df_results_rq1.head(5).to_string(index=False))
else:
    print("未检测到符合定义的 Lone Wolf 行为。")

# ==============================================================================
# RQ2: 独狼行为的影响 (Impact Analysis - Merge Rate)
# ==============================================================================
print("\n" + "="*80)
print(">>> [RQ2 Analysis] 独狼行为对合并率的影响")
print("="*80)

target_population = active_collab_env.copy()

# 检查必要的列是否存在 (注意后缀 _ai)
needed_cols = ['state_ai', 'merged_at_ai']
if not all(col in target_population.columns for col in needed_cols):
     print(f"错误：无法执行 RQ2 分析，因为数据中缺少必要列: {needed_cols}。请检查源数据或标准化步骤。")
else:
    # 打标签：是否为 Lone Wolf
    target_population['is_lone_wolf'] = ~target_population['pr_id_ai'].isin(communicated_pr_ids)

    # 去重并提取所需列
    analysis_df = target_population[['pr_id_ai', 'state_ai', 'merged_at_ai', 'is_lone_wolf']].drop_duplicates()

    # 【关键修正 2】：更新合并率计算逻辑
    def calculate_metrics(df, group_name):
        total = len(df)
        if total == 0:
            print(f"[{group_name}] 样本量不足。")
            return 0, 0
        
        # 新逻辑：state 为 'closed' 且 merged_at 不为空
        merged_condition = (df['state_ai'] == 'closed') & (df['merged_at_ai'].notna())
        merged_count = len(df[merged_condition])
        
        merge_rate = (merged_count / total) * 100
        print(f"[{group_name}] 总数: {total:<6} | 合并数: {merged_count:<6} | 合并率: {merge_rate:.2f}%")
        return merged_count, total - merged_count

    print("-" * 60)
    print("对比结果 (Group A: 独狼 AI  vs  Group B: 社交 AI)")
    print("合并定义: state == 'closed' AND merged_at is not null")
    print("-" * 60)
    merged_a, rejected_a = calculate_metrics(analysis_df[analysis_df['is_lone_wolf'] == True], "Group A (独狼)")
    merged_b, rejected_b = calculate_metrics(analysis_df[analysis_df['is_lone_wolf'] == False], "Group B (社交)")
    print("-" * 60)

    # 统计检验
    if (merged_a + rejected_a > 0) and (merged_b + rejected_b > 0):
        contingency = [[merged_a, rejected_a], [merged_b, rejected_b]]
        try:
            chi2, p, _, _ = chi2_contingency(contingency)
            print(f"统计检验 (Chi-square): p-value = {p:.4e}")
            if p < 0.05:
                print("结论：两组的合并率存在【统计学显著】差异。")
            else:
                print("结论：两组合并率差异不显著。")
        except ValueError:
             print("无法进行统计检验(样本可能过少)。")


# ==============================================================================
# RQ3: Agent 行为画像 (Habitual Analysis)
# ==============================================================================
print("\n" + "="*80)
print(">>> [RQ3 Analysis] Agent 的独狼倾向性分析")
print("="*80)

# 确保 analysis_df 存在且 agent 列存在
if 'analysis_df' in locals() and not analysis_df.empty and 'agent' in df_ai_clean.columns:
    # 映射 agent 名称
    pr_agent_map = df_ai_clean.set_index('pr_id')['agent'].to_dict()
    analysis_df['agent_name'] = analysis_df['pr_id_ai'].map(pr_agent_map)
    analysis_df['agent_name'] = analysis_df['agent_name'].fillna('Unknown_Agent')

    # 按 Agent 分组统计
    agent_stats = analysis_df.groupby('agent_name').agg(
        total_collab_prs=('is_lone_wolf', 'count'),
        lone_wolf_prs=('is_lone_wolf', 'sum')
    ).reset_index()

    # 计算独狼率
    agent_stats['lw_rate'] = (agent_stats['lone_wolf_prs'] / agent_stats['total_collab_prs']) * 100

    # 排序并展示 (只看协作 PR 数 >= 5 的活跃 Agent)
    min_prs = 10
    active_agents = agent_stats[agent_stats['total_collab_prs'] >= min_prs].sort_values('lw_rate', ascending=False)

    print(f"Agent 独狼倾向排行榜 (仅展示协作 PR 数 >= {min_prs} 的 Agent):")
    print("-" * 80)
    if not active_agents.empty:
        pd.set_option('display.max_rows', None)
        # 格式化输出 lw_rate
        print(active_agents.to_string(index=False, formatters={'lw_rate': '{:.1f}%'.format}))
        
        high_risk = len(active_agents[active_agents['lw_rate'] > 80])
        print("-" * 80)
        print(f"分析：在 {len(active_agents)} 个活跃 Agent 中，有 {high_risk} 个表现出极高(>80%)的独狼倾向。")
    else:
        print(f"没有 Agent 的协作 PR 数达到 {min_prs} 个。")
else:
    print("无法进行 RQ3 分析（缺少必要数据或列）。")
```
